#Day 24: Exploring Loss Functions

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# Generating a simple dataset for regression
X_train = np.random.rand(100, 1)  # 100 samples, 1 feature
y_train = 3 * X_train + np.random.randn(100, 1) * 0.1  # y = 3x + noise
```

```python
# Building a simple linear regression model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, input_shape=(1,))
])

# Visualizing the dataset
plt.scatter(X_train, y_train, color='blue')
plt.title('Training Data')
plt.xlabel('Input Feature')
plt.ylabel('Target')
plt.show()
```

```python
# Compile the model using Mean Squared Error loss function
model.compile(optimizer='adam', loss='mean_squared_error')

# Train the model
history = model.fit(X_train, y_train, epochs=50, verbose=0)

# Plotting the loss curve
plt.plot(history.history['loss'], label='Training Loss (MSE)')
plt.title('Training Loss Curve with Mean Squared Error')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Binary Crossentropy loss function example
# Generating synthetic binary classification data
X_class = np.random.rand(100, 2)
y_class = np.random.randint(0, 2, size=(100, 1))

# Building a binary classification model
model_class = tf.keras.Sequential([
    tf.keras.layers.Dense(1, activation='sigmoid', input_shape=(2,))
])

# Compiling the model with Binary Crossentropy
model_class.compile(optimizer='adam', loss='binary_crossentropy')

# Train the model
history_class = model_class.fit(X_class, y_class, epochs=50, verbose=0)

# Plotting the loss curve
plt.plot(history_class.history['loss'], label='Training Loss (Binary Crossentropy)')
plt.title('Training Loss Curve with Binary Crossentropy')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Categorical Crossentropy example for multi-class classification
# Generating synthetic multi-class data
X_multi = np.random.rand(150, 3)
y_multi = tf.keras.utils.to_categorical(np.random.randint(0, 3, size=(150,)))

# Building a multi-class classification model
model_multi = tf.keras.Sequential([
    tf.keras.layers.Dense(3, activation='softmax', input_shape=(3,))
])

# Compiling the model with Categorical Crossentropy
model_multi.compile(optimizer='adam', loss='categorical_crossentropy')

# Train the model
history_multi = model_multi.fit(X_multi, y_multi, epochs=50, verbose=0)

# Plotting the loss curve
plt.plot(history_multi.history['loss'], label='Training Loss (Categorical Crossentropy)')
plt.title('Training Loss Curve with Categorical Crossentropy')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Hinge Loss example for Support Vector Machines (SVM)
# Generating synthetic binary classification data
X_hinge = np.random.rand(100, 2)
y_hinge = np.random.choice([-1, 1], size=(100,))

# Building an SVM model using Hinge Loss
model_hinge = tf.keras.Sequential([
    tf.keras.layers.Dense(1, activation='tanh', input_shape=(2,))
])

# Compiling the model with Hinge loss
model_hinge.compile(optimizer='adam', loss='hinge')

# Train the model
history_hinge = model_hinge.fit(X_hinge, y_hinge, epochs=50, verbose=0)

# Plotting the loss curve
plt.plot(history_hinge.history['loss'], label='Training Loss (Hinge Loss)')
plt.title('Training Loss Curve with Hinge Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Compiling the model with MAE loss
model.compile(optimizer='adam', loss='mean_absolute_error')

# Training the model with MAE loss
history_mae = model.fit(X_train, y_train, epochs=50, verbose=0)

# Plotting the loss curve for MAE
plt.plot(history_mae.history['loss'], label='MAE Loss')
plt.title('Training Loss Curve (Mean Absolute Error)')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# RMSE requires using a custom loss function since TensorFlow does not provide it directly
def rmse(y_true, y_pred):
    return tf.sqrt(tf.reduce_mean(tf.square(y_true - y_pred)))

# Compiling the model with RMSE as the loss function
model.compile(optimizer='adam', loss=rmse)

# Training the model with RMSE
history_rmse = model.fit(X_train, y_train, epochs=50, verbose=0)

# Plotting the loss curve for RMSE
plt.plot(history_rmse.history['loss'], label='RMSE Loss')
plt.title('Training Loss Curve (Root Mean Squared Error)')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

##Loss functions are key to optimizing model performance. Different types of problems require different loss functions, and understanding them allows you to choose the best one for your task.
