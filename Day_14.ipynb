##Day 14 - Finding Determinant and Inverse of np array and tensors




1. Finding the Determinant of a Matrix


```python
import numpy as np

# Example: 2x2 matrix
matrix = np.array([[1, 2], [3, 4]])

# Calculate the determinant
det = np.linalg.det(matrix)
print("Determinant:", det)
```

```python
import tensorflow as tf

# Convert the NumPy matrix to a TensorFlow tensor
tensor_matrix = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)

# Calculate the determinant using TensorFlow
det_tf = tf.linalg.det(tensor_matrix)
print("Determinant (TensorFlow):", det_tf.numpy())
```

2. Finding the Inverse of a Matrix


```python
# Calculate the inverse using NumPy
inverse = np.linalg.inv(matrix)
print("Inverse Matrix:\n", inverse)
```

```python
# Calculate the inverse using TensorFlow
inverse_tf = tf.linalg.inv(tensor_matrix)
print("Inverse Matrix (TensorFlow):\n", inverse_tf.numpy())
```

3. Working with Tensors


```python
# Example: 3x3 matrix (tensor)
tensor_matrix_3x3 = tf.constant([[1, 2, 3], [0, 1, 4], [5, 6, 0]], dtype=tf.float32)

# Calculate the determinant
det_tensor = tf.linalg.det(tensor_matrix_3x3)
print("Determinant (Tensor TensorFlow):", det_tensor.numpy())
```

```python
# Calculate the inverse
inverse_tensor = tf.linalg.inv(tensor_matrix_3x3)
print("Inverse Matrix (Tensor TensorFlow):\n", inverse_tensor.numpy())
```
