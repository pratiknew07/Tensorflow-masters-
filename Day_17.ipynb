# Day 17: Deep Dive into Activation Functions in TensorFlow 🚀Section

**1. Introduction to Activation Functions**
Activation functions decide whether a neuron should be activated or not, bringing non-linearity into the model and enabling it to handle complex data.

```python
import tensorflow as tf

# Sample input tensor
x = tf.constant([-10.0, -1.0, 0.0, 1.0, 10.0])
```

**2. Exploring the Sigmoid Activation Function** **bold text**

The Sigmoid function maps input values to a range between 0 and 1, making it useful for binary classification.

```python
# Sigmoid Activation Function
sigmoid = tf.nn.sigmoid(x)
print("Sigmoid:", sigmoid.numpy())
```

**3. Exploring the Tanh Activation Function**

The Tanh function maps input values to a range between -1 and 1, making it zero-centered, which helps in optimization.

```python
# Tanh Activation Function
tanh = tf.nn.tanh(x)
print("Tanh:", tanh.numpy())
```

**4. Exploring the ReLU Activation Function**

ReLU (Rectified Linear Unit) is widely used because of its simplicity and effectiveness in avoiding the vanishing gradient problem

```python
# ReLU Activation Function
relu = tf.nn.relu(x)
print("ReLU:", relu.numpy())
```

**5. Visualizing Activation Functions**

Let's visualize how these activation functions behave over a range of input values.

```python
import matplotlib.pyplot as plt
import numpy as np

# Define a range of values
x_values = np.linspace(-10, 10, 100)

# Compute activation values
sigmoid_values = tf.nn.sigmoid(x_values).numpy()
tanh_values = tf.nn.tanh(x_values).numpy()
relu_values = tf.nn.relu(x_values).numpy()

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(x_values, sigmoid_values, label="Sigmoid")
plt.plot(x_values, tanh_values, label="Tanh")
plt.plot(x_values, relu_values, label="ReLU")
plt.title("Activation Functions")
plt.xlabel("Input")
plt.ylabel("Output")
plt.legend()
plt.grid(True)
plt.show()
```

**6. Using Activation Functions in a Model**

Here’s how you can apply these activation functions in a simple neural network model.




```python
# Simple neural network model using activation functions
model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(5, activation='tanh'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Summarizing the model
model.summary()
```
