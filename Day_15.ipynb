#Day 15 - working with tensors transpose and other operations

1. Transpose of a Tensor
Why It Matters: Transposing a tensor reorders its dimensions, which is particularly useful in scenarios like image processing, where you might need to switch rows and columns, or in deep learning, where tensor shapes need to align across different layers.

```python
import tensorflow as tf

# Creating a 2x3 tensor
tensor = tf.constant([[1, 2, 3], [4, 5, 6]])

# Transposing the tensor (switching rows and columns)
transposed_tensor = tf.transpose(tensor)
print("Original Tensor:\n", tensor.numpy())
print("Transposed Tensor:\n", transposed_tensor.numpy())
```

2. Reshaping a Tensor
Why It Matters: Reshaping allows you to change the dimensions of a tensor without altering its data, which is vital for feeding data into neural networks where specific input shapes are required.

```python
# Reshape the tensor from 2x3 to 3x2
reshaped_tensor = tf.reshape(tensor, [3, 2])
print("Reshaped Tensor:\n", reshaped_tensor.numpy())
```

3. Expanding and Squeezing Dimensions
Why It Matters: Sometimes, we need to add or remove dimensions from tensors to ensure compatibility with different operations or models

```python
# Expanding the tensor to add an extra dimension
expanded_tensor = tf.expand_dims(tensor, axis=0)  # Adding a new dimension at axis 0
print("Expanded Tensor Shape:", expanded_tensor.shape)
```

```python
# Squeezing the tensor to remove single-dimensional entries
squeezed_tensor = tf.squeeze(expanded_tensor)
print("Squeezed Tensor Shape:", squeezed_tensor.shape)
```

4. Permuting Dimensions
Why It Matters: Permuting allows you to reorder the dimensions of a tensor, which is essential in certain machine learning tasks where the arrangement of data needs to be changed.

```python
# Permute the dimensions of a 3D tensor
tensor_3d = tf.random.uniform([2, 3, 4])
permuted_tensor = tf.transpose(tensor_3d, perm=[1, 0, 2])
print("Original Tensor Shape:", tensor_3d.shape)
print("Permuted Tensor Shape:", permuted_tensor.shape)
```
